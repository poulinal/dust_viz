```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from PIL import Image

#take tiff file and convert to imshow
def plot_tiff_as_image(tiff_file):
    # Read the TIFF file using pandas
    df = Image.open(tiff_file)
    
    # Convert the DataFrame to a NumPy array
    data = np.array(df)
    
    figure, axes = plt.subplots(figsize=(10, 10))  # Create a figure and axes with a specific size
    # Plot the data as an image
    imshowed= axes.imshow(data, cmap='gray', interpolation='nearest')
    axes.axis('off')  # Hide the axes
    figure.show()
    print(imshowed.get_clim())
    
# Example usage
if __name__ == "__main__":
    tiff_file = '/Users/alexpoulin/Library/CloudStorage/OneDrive-NortheasternUniversity/DeLTA_Lab/data/10.10.2024/ARPES_MnTe(exp30)_Ekinsweep_241010_175544/ARPES_MnTe(exp30)_Ekinsweep_241010_175545/Sum/ARPES_MnTe(exp30)_Ekinsweep_SUM_001.TIF'  # Replace with your TIFF file path
    plot_tiff_as_image(tiff_file)
```
